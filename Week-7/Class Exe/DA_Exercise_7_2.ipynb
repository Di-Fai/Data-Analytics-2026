This notebook uses Pandas to analyze employee sales performance data for RetailHub Inc. It covers DataFrame creation, data inspection, filtering, sorting, calculated columns, conditional ratings, and GroupBy analysis.

## 1. Import Pandas and Create the DataFrame

In this section, I import Pandas and create a DataFrame using employee sales records.

```python
import pandas as pd

data = {
    "name": ["Alice", "Ben", "Carmen", "Dev", "Eve", "Frank", "Grace", "Hiro"],
    "city": ["London", "Manchester", "London", "Leeds", "Birmingham", "Manchester", "Leeds", "London"],
    "sales": [4200.0, 3100.0, 5050.0, 2800.0, 1950.0, 3750.0, 4600.0, 3300.0],
    "months": [12, 6, 24, 3, 18, 9, 15, 7]
}

df = pd.DataFrame(data)

print("Employee DataFrame:")
print(df)

print("\nShape:")
print(df.shape)

print("\nData types:")
print(df.dtypes)

print("\nInfo:")
df.info()
```

## 2. Inspect and Describe the Data

In this section, I use Pandas methods to preview the data and calculate summary statistics.

```python
print("First 3 rows:")
print(df.head(3))

print("\nLast 3 rows:")
print(df.tail(3))

print("\nSummary statistics:")
print(df.describe())

print(f"\nMean sales: £{df['sales'].mean():,.2f}")
print(f"Max sales: £{df['sales'].max():,.2f}")
```

## 3. Select, Filter, and Sort

In this section, I select specific columns, filter employees by sales, and sort the DataFrame from highest to lowest sales.

```python
name_sales = df[["name", "sales"]]

print("Name and sales columns:")
print(name_sales)

high_sales = df[df["sales"] > 3500.0]

print("\nEmployees with sales greater than £3,500:")
print(high_sales)

df_sorted = df.sort_values("sales", ascending=False)

print("\nDataFrame sorted by sales from highest to lowest:")
print(df_sorted)

print("\nTop-selling employee record:")
print(df_sorted.iloc[0])
```

## 4. Calculated Column and Conditional Rating

In this section, I add an 8% commission column and create a rating column based on each employee's sales performance.

```python
df["commission"] = df["sales"] * 0.08

df["rating"] = df["sales"].apply(
    lambda x: "Star" if x >= 4000
    else ("Good" if x >= 2500 else "Developing")
)

print("Updated DataFrame:")
print(df)

print("\nCommission per employee:")
for _, row in df.iterrows():
    print(f"{row['name']}: £{row['commission']:,.2f}")
```

## 5. GroupBy — Sales by City

In this section, I use GroupBy to summarize total sales, average sales, and employee count by city.

```python
city_totals = df.groupby("city")["sales"].sum()

print("Total sales by city:")
print(city_totals)

city_means = df.groupby("city")["sales"].mean()

print("\nMean sales by city:")
print(city_means)

city_summary = df.groupby("city").agg({"sales": ["sum", "mean", "count"]})

print("\nFull city summary:")
print(city_summary)

top_city = city_totals.idxmax()

print(f"\nTop city: {top_city}")
print(f"Total sales for {top_city}: £{city_totals[top_city]:,.2f}")
```

## Summary

London is the top-performing city because it has the highest total sales at £12,550.00. The GroupBy results show that London also has the strongest average sales compared to the other cities, while Birmingham has the lowest total because it only has one employee in the data. Overall, the data shows that employees in London are performing strongly, especially Carmen, who is the top-selling employee.
