# Exercise 2.A Fundamentals of pandas

```python
import numpy as np
import pandas as pd
```

## Series

```python
aa = [1, 2, 3, 4]

ab = (5, 6, 7, 8)

ac = {9, 10, 11, 12}

ad = {
    "first": 13,
    "second": 14,
    "third": 15,
    "fourth": 16
}
```

```python
ser1 = pd.Series(aa)

ser2 = pd.Series(ab)

# A set is unordered, so I converted it to a sorted list first.
ser3 = pd.Series(sorted(ac))

ser4 = pd.Series(ad)

print("ser1:")
print(ser1)

print("\nser2:")
print(ser2)

print("\nser3:")
print(ser3)

print("\nser4:")
print(ser4)
```

```python
try:
    ser5 = pd.Series(17, 18, 19, 20)

except TypeError as error:
    print(error)

# TypeError message:
# Index(...) must be called with a collection of some kind, 18 was passed
```

## DataFrames from Series

```python
df1 = pd.DataFrame(ser1, columns=["ser1"])

df1
```

```python
df1.insert(1, "ser2", ser2)

df1.insert(2, "ser3", ser3)

df1
```

```python
df1["ser4"] = ser4

df1

# ser4 is different because it was created from a dictionary.
# The dictionary created labels like "first", "second", "third", and "fourth".
# df1 uses number indexes like 0, 1, 2, and 3.
# Since the indexes do not match, pandas fills the ser4 column with NaN.
```

```python
df1.loc[4] = [5, 9, 12, 0]

df1
```

Looking at the info results, I can determine the number of rows and columns in the DataFrame. I can also see the column names, the data types, and whether any columns have missing values. This helps me understand the structure of the data before analyzing it.

Looking at the describe results, I can see useful statistics like count, mean, minimum, maximum, and standard deviation for numeric columns. These are useful for understanding the spread of the data. Some summaries may not be as useful if the numbers are ID values, because IDs are labels and not measurements.

## Coding a DataFrame

```python
test_data = {
    "ID": [1, 2, 3, 4, 5, 6, 7, 8],
    "Name": ["Alicia", "Bob", "Charlie", "Dominic", "Eve", "Frank", "Grace", "Heidi"],
    "Age": [25, np.nan, 22, 19, 31, 35, np.nan, 18],
    "City": ["NY", "LA", "Chicago", "Houston", "Phoenix", "Boston", "Austin", "San Diego"],
    "Score": [85, 92, -1, 65, 78, np.nan, 55, 90]
}
```

```python
df4 = pd.DataFrame(test_data)

df4
```

```python
df4.to_csv("test_data.csv", index=False)
```
