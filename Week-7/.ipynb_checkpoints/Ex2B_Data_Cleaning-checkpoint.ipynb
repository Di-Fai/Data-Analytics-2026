## Modifying a DataFrame

```python
df_cleaned = df.drop(df[df["Score"] < 0].index).copy()

df_cleaned

# This removes rows where the Score is less than 0.

# The score -1 is invalid, so that row is removed.
```

```python
df_cleaned = df_cleaned.drop(columns=["City"])

df_cleaned

# This removes the City column from the DataFrame.
```

```python
mean_age = df_cleaned["Age"].mean()

df_cleaned["Age"] = df_cleaned["Age"].fillna(mean_age)

df_cleaned

# This calculates the average age, then replaces all missing Age values with that average
```

```python
mean_score = df_cleaned["Score"].mean()

df_cleaned["Score"] = df_cleaned["Score"].replace(0, mean_score)

df_cleaned

# This replaces any Score value of 0 with the average score.
# In this dataset, there may not be a 0, so the DataFrame may look the same.
```

```python
df_cleaned = df_cleaned.dropna()

df_cleaned

# This removes any rows that still contain missing values.

# For example, if a row still has a missing Score, it will be removed.
```

```python
if "ID" in df_cleaned.columns:
    df_cleaned = df_cleaned.set_index("ID")

df_cleaned

# This makes the ID column become the index labels.

# After this, ID is no longer a regular column. It becomes the row index.
```
