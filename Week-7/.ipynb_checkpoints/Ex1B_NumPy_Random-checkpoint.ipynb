# Exercise 1.B NumPy Random

```python
import numpy as np
```

## Generating Random Numbers

```python
r_1 = np.random.randint(1, 21, size=2)

print(r_1)
```

```python
r_2 = np.random.randint(1, 21, size=(1, 2))

print(r_2)
```

```python
r_3 = np.random.rand(2, 1)

print(r_3)
```

```python
r_4 = np.random.randint(20, 101, size=(4, 5))

print(r_4)
```

```python
r_5 = np.random.rand(4, 5)

print(r_5)
```

## Generating Random Choices from an Array

```python
students = np.array([
    "Chelsea", "Alex", "Brian", "Maria",
    "James", "Aaliyah", "David", "Sophia",
    "Michael", "Nadia", "Chris", "Tiana"
])

print(students)
```

```python
call_on = np.random.choice(students)

print(call_on)
```

```python
breakout_groups = np.random.choice(students, size=(3, 4))

print(breakout_groups)

# This may not be the best way to assign breakout groups because the same student
# can be selected more than once. That means one person could appear in multiple
# groups while another student may not be selected at all.
```

```python
call_on_weighted = np.random.choice(
    students,
    size=5,
    p=[0.30, 0.10, 0.05, 0.08, 0.12, 0.00, 0.07, 0.06, 0.05, 0.04, 0.08, 0.05]
)

print(call_on_weighted)

# The p values control how likely each student is to be selected.
# Chelsea has a 30% chance, so she is the most likely to be selected.
# Aaliyah has a 0% chance, so she cannot be selected.
# The other students have smaller probabilities, and all probabilities add up to 1.00.
```

## Generating a Normal Distribution

```python
bell_curve = np.random.normal(loc=10, scale=5, size=(4, 5))

print(bell_curve)
```
