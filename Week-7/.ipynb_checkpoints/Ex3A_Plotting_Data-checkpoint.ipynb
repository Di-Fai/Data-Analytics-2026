# Exercise 3.A Getting familiar with Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Creating Basic Plots

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]
```

```python
plt.plot(x, y)
plt.show()
```

```python
plt.plot(x, y, marker='o')
plt.show()
```

```python
plt.plot(x, y, marker='o')

plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')

plt.show()
```

## Styling Plots

```python
plt.plot(x, y, linestyle='--', color='red')

plt.title('Dashed Red Line Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')

plt.show()
```

```python
plt.plot(
    x,
    y,
    marker='s',
    markersize=10,
    markerfacecolor='green',
    markeredgecolor='green'
)

plt.title('Graph with Square Green Markers')
plt.xlabel('X axis')
plt.ylabel('Y axis')

plt.show()
```

```python
plt.plot(
    x,
    y,
    marker='^',
    markersize=10,
    markerfacecolor='blue',
    markeredgecolor='blue',
    color='black'
)

plt.title('Graph with Triangle Blue Markers')
plt.xlabel('X axis')
plt.ylabel('Y axis')

plt.show()
```

```python
plt.plot(
    x,
    y,
    marker='D',
    markersize=10,
    markerfacecolor='purple',
    markeredgecolor='purple',
    color='orange'
)

plt.title('Graph with Diamond Purple Markers')
plt.xlabel('X axis')
plt.ylabel('Y axis')

plt.show()
```

## Conclusion

In this exercise, I learned how to create basic plots using Matplotlib. I practiced creating x and y data points, plotting a line graph, adding markers, and customizing the graph with titles and axis labels. I also learned how to style plots by changing line style, line color, marker shape, marker size, and marker color. This helped me understand how visual formatting makes data easier to read and explain.


# Exercise 3.B More plotting practice

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Advanced Plotting Techniques

```python
products = ['Product A', 'Product B', 'Product C', 'Product D']
sales = [23, 17, 35, 29]

plt.bar(products, sales)
plt.show()
```

```python
products = ['Product A', 'Product B', 'Product C', 'Product D']
sales = [23, 17, 35, 29]

plt.bar(products, sales, color='skyblue', width=0.5)

plt.title('Product Sales')
plt.xlabel('Products')
plt.ylabel('Sales')

plt.show()
```

```python
x = np.random.rand(50)
y = np.random.rand(50)

plt.scatter(x, y)
plt.show()
```

```python
x = np.random.rand(50)
y = np.random.rand(50)

sizes = np.random.rand(50) * 100
colors = np.random.rand(50)

plt.scatter(x, y, s=sizes, c=colors, cmap='viridis')

plt.title('Customized Scatter Plot')
plt.xlabel('X Values')
plt.ylabel('Y Values')

plt.colorbar(label='Color Scale')

plt.show()
```

## Subplots

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)

random_numbers = np.random.randn(1000)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# First subplot: sine wave line chart
axes[0].plot(x, y)
axes[0].set_title('Sine Wave')
axes[0].set_xlabel('X Values')
axes[0].set_ylabel('sin(x)')

# Second subplot: histogram
axes[1].hist(random_numbers, bins=30)
axes[1].set_title('Histogram of Random Numbers')
axes[1].set_xlabel('Value')
axes[1].set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```

## Conclusion

In this exercise, I practiced creating more advanced plots with Matplotlib. I learned how to build and customize bar charts, create scatter plots with different marker sizes and colors, and use colormaps to add more meaning to a visualization. I also learned how to create subplots, which allows multiple charts to appear in one figure. This helped me understand how Matplotlib can be used to make data easier to compare, explain, and present.
