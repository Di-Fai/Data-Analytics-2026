## Sample Arrays

```python
r_1 = np.random.randint(1, 21, size=2)

r_2 = np.random.randint(1, 21, size=(1, 2))

r_3 = np.random.random(size=(2, 1))

r_4 = np.random.randint(20, 51, size=(3, 5))

r_5 = np.random.randint(1, 100, size=5)

print("r_1:")
print(r_1)

print("\nr_2:")
print(r_2)

print("\nr_3:")
print(r_3)

print("\nr_4:")
print(r_4)

print("\nr_5:")
print(r_5)
```

## Permutations

```python
print("Current value of r_4:")
print(r_4)
```

```python
np.random.shuffle(r_4)

print("r_4 after using shuffle:")
print(r_4)
```

The shuffle function changed the order of the rows in r_4 directly. It did not create a new array. It changed the original r_4 array.

```python
new_shuffle = np.random.permutation(r_4)

print("new_shuffle:")
print(new_shuffle)

print("\nr_4 after using permutation:")
print(r_4)
```

I would use shuffle when I want to change the original array directly. I would use permutation when I want to create a shuffled copy but keep the original array the same.

## Mathematical Operations

```python
print("r_1:")
print(r_1)
print("Shape of r_1:", r_1.shape)

print("\nr_2:")
print(r_2)
print("Shape of r_2:", r_2.shape)

result = np.add(r_1, r_2)

print("\nr_1 + r_2:")
print(result)
print("Shape of result:", result.shape)
```

```python
print("r_1:")
print(r_1)
print("Shape of r_1:", r_1.shape)

print("\nr_3:")
print(r_3)
print("Shape of r_3:", r_3.shape)

result = np.add(r_1, r_3)

print("\nr_1 + r_3:")
print(result)
print("Shape of result:", result.shape)
```

```python
print("r_2:")
print(r_2)
print("Shape of r_2:", r_2.shape)

print("\nr_4:")
print(r_4)
print("Shape of r_4:", r_4.shape)

try:
    result = np.add(r_2, r_4)
    print("\nr_2 + r_4:")
    print(result)
    print("Shape of result:", result.shape)

except ValueError as error:
    print("\nr_2 + r_4 gives an error:")
    print(error)
```

```python
try:
    print("r_1 + r_4:")
    print(np.add(r_1, r_4))

except ValueError as error:
    print("r_1 + r_4 gives an error:")
    print(error)


try:
    print("\nr_2 + r_4:")
    print(np.add(r_2, r_4))

except ValueError as error:
    print("\nr_2 + r_4 gives an error:")
    print(error)


try:
    print("\nr_3 + r_4:")
    print(np.add(r_3, r_4))

except ValueError as error:
    print("\nr_3 + r_4 gives an error:")
    print(error)

# These errors happen because the shapes of the arrays do not match correctly for broadcasting.
```

```python
print("r_4 + np.array([1]):")
print(np.add(r_4, np.array([1])))
```

```python
print("r_4 + np.array([[1]]):")
print(np.add(r_4, np.array([[1]])))
```

```python
print("r_4 + np.array([1, 1, 1, 1, 1]):")
print(np.add(r_4, np.array([1, 1, 1, 1, 1])))
```

```python
try:
    print("r_4 + np.array([[1], [1]]):")
    print(np.add(r_4, np.array([[1], [1]])))

except ValueError as error:
    print("This gives an error:")
    print(error)

# Error reason:
# r_4 has 3 rows and 5 columns.
# np.array([[1], [1]]) has 2 rows and 1 column.
# The row sizes do not match, so NumPy cannot broadcast them together.
```

Broadcasting is when NumPy tries to stretch a smaller array so it can work with a larger array. When using the add function, NumPy adds values based on matching positions. If one array has a shape that can fit across the other array, NumPy will add it. The shape of the result is usually based on the larger array.

## Grocery Problem

```python
# Order: vegetables, fruits, snacks, beverages

bag1 = np.array([2, 3, 4, 0])
bag2 = np.array([4, 0, 1, 2])

print("Bag 1:")
print(bag1)

print("\nBag 2:")
print(bag2)
```

```python
bag1_total = np.sum(bag1)
bag2_total = np.sum(bag2)

print(f"The first bag contains {bag1_total} items, and the second bag contains {bag2_total} items.")
```

```python
category_totals = bag1 + bag2

print(
    f"There are {category_totals[0]} vegetables, "
    f"{category_totals[1]} fruits, "
    f"{category_totals[2]} boxes of snacks, "
    f"and {category_totals[3]} beverages in the two grocery bags."
)
```

## Teamwork

```python
ali = np.array([5, 4, 6, 4, 2])
```

```python
ben = np.array([0, 5, 5, 3, 4])
```

```python
carla = np.array([3, 6, 0, 3, 4])
```

```python
all_knocks = np.array([ali, ben, carla])

print("All knocks:")
print(all_knocks)
```

```python
person_totals = np.sum(all_knocks, axis=1)
group_total = np.sum(all_knocks)

print("Total doors knocked by each person:")
print("Ali:", person_totals[0])
print("Ben:", person_totals[1])
print("Carla:", person_totals[2])

print("\nTotal doors knocked by the group:")
print(group_total)
```

```python
daily_totals = np.sum(all_knocks, axis=0)

rolling_total_doors = np.cumsum(daily_totals)

estimated_votes = rolling_total_doors * 0.5

print("Daily totals:")
print(daily_totals)

print("\nRolling total of doors knocked:")
print(rolling_total_doors)

print("\nRolling total of estimated votes:")
print(estimated_votes)
```
