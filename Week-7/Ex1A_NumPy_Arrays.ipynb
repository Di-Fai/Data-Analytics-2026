# Exercise 1.A NumPy: Manipulating Arrays

```python
import numpy as np
```

## Looping Through Arrays

```python
arrA = np.array([
    [11, 12, 13],
    [14, 15, 16]
])

print(arrA)
```

```python
for x in arrA:
    print(x * 2)
```

```python
for x in arrA:
    for y in x:
        print(y * 2)
```

The first loop goes through each row of the array, so the result prints each full row multiplied by 2. The second loop goes inside each row and works with each individual number one at a time. That is why the first loop prints arrays, while the second loop prints single numbers.

```python
for x in arrA:
    for y in x:
        loop_result = x * 2
        print(loop_result)

# This result is different because x represents the whole row.
# Even though y loops through each individual number, the code is multiplying x, not y.
# That means each row gets printed multiple times.
```

## Joining and Stacking Arrays

```python
joined_array_1 = np.concatenate((arrA, arrA), axis=0)

print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA, arrA), axis=1)

print(joined_array_2)
```

```python
stacked_array_1 = np.stack((joined_array_1, joined_array_1))

print(stacked_array_1)
```

```python

stacked_array_2 = np.stack((joined_array_1, joined_array_2))

print(stacked_array_2)

# This code gives an error because joined_array_1 and joined_array_2 have different shapes.
# joined_array_1 has 4 rows and 3 columns.
# joined_array_2 has 2 rows and 6 columns.
# NumPy cannot stack arrays unless they have the same shape.
```

```python
another_joined_array = np.concatenate((joined_array_1, arrA))

print(another_joined_array)
```

```python

another_2 = np.concatenate((joined_array_1, arrA), axis=1)

print(another_2)

# This code gives an error because axis=1 joins side by side.
# For that to work, the arrays must have the same number of rows.
# joined_array_1 has 4 rows, but arrA has only 2 rows.
```

```python
h_stack = np.hstack((arrA, arrA))
# hstack joins the arrays horizontally, side by side.

v_stack = np.vstack((arrA, arrA))
# vstack joins the arrays vertically, one on top of the other.

d_stack = np.dstack((arrA, arrA))
# dstack stacks the arrays along a third dimension, adding depth.

print("hstack result:")
print(h_stack)

print("\nvstack result:")
print(v_stack)

print("\ndstack result:")
print(d_stack)
```

## Splitting Arrays

```python
splitsky = np.array_split(joined_array_1, 2)

print(splitsky)

# The result is 2 arrays.
# Each array is two-dimensional.
# Each array has 2 rows and 3 columns.
```

```python
spliterino = np.array_split(joined_array_1, 3)

print(spliterino)

# The result is 3 arrays.
# Each array is still two-dimensional.
# Because 4 rows cannot split evenly into 3 groups, NumPy makes uneven groups.
# The first array has 2 rows and 3 columns.
# The second and third arrays each have 1 row and 3 columns.
```

## Searching and Replacing Array Values

```python
result_1 = np.where(arrA % 2 == 0, arrA, "odd")

print(result_1)

# This checks each number in arrA. If the number is even, it prints the number. If the number is not even, it prints "odd".
```

## Sorting and Filtering Arrays

```python
arrB = np.array([
    [7, 6],
    [13, 5],
    [7, 3]
])

arrB_sorted = np.sort(arrB)

print(arrB_sorted)
```

```python
print(arrB)

z = arrB < 7

arrC = arrB[z]

print(arrC)
```

```python
print(arrB)

y = arrB >= 7

arrD = arrB[y]

print(arrD)
```
