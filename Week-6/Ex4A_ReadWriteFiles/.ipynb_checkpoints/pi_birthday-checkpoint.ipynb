# Pi Birthday Checker

This notebook checks whether a six-digit birthday appears in the first million digits of pi.
The birthday should be entered as a string in MMDDYY format, such as "070281".

```python
pi = open("pi_million_digits.txt", "r")

print(pi.readline())

pi.close()
```

```python
pi = open("pi_million_digits.txt", "r")

pi_lines = pi.readlines()

pi.close()
```

```python
def pi_bday_check(bday):
    for i in pi_lines:
        if bday in i:
            print("Your birthday is in the first million digits of pi")
            break
```

```python
def pi_bday_check2(bday):
    birthday_found = 0

    for i in pi_lines:
        if bday in i:
            print("Your birthday is in the first million digits of pi")
            birthday_found += 1
            break

    if birthday_found != 1:
        print("Sorry, your birthday was not found in the first million digits of pi")
```

```python
pi_string = ""

for i in pi_lines:
    pi_string += i.strip()

print(pi_string[0:300])
```

```python
birthday_position = pi_string.index("070281") - 1

# The -1 is used because Python indexing starts at 0,
# and pi_string starts with 3 and the decimal point.
# We subtract 1 to get the birthday's decimal place position.

print(birthday_position)
```

```python
def pi_bday_check3(bday):
    birthday_found = 0
    bday_position = ""

    for i in pi_lines:
        if bday in i:
            print("Your birthday is in the first million digits of pi")
            birthday_found += 1
            break

    if birthday_found != 1:
        print("Sorry, your birthday was not found in the first million digits of pi")
    else:
        bday_position = pi_string.index(bday) - 1

        # The -1 is used because Python indexing starts at 0,
        # and pi_string starts with 3 and the decimal point.

        print(f"Your birthday begins at decimal place {bday_position}")
```

```python
pi_bday_check3("070281")
pi_bday_check3("100799")
pi_bday_check3("081505")
pi_bday_check3("110252")
```

This notebook opens a file that contains the first million digits of pi. First, it reads the file line by line. Then it checks if a six-digit birthday appears anywhere in those lines. After that, it combines all the lines into one long string so Python can find the exact decimal position where the birthday starts.
