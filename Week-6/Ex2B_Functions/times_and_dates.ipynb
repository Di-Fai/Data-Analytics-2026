### This is Lab 2 of Exercise 2.B. In this notebook, I am working with datetime functions in Python.

```python
import datetime
```

# Today

```python
today = datetime.datetime.now()
```

```python
print("Today is " + today.strftime("%A, %B %d, %Y"))
```

# My birthday

```python
birthday = datetime.date(1998, 6, 14)
```

```python
print("My birthday is " + birthday.strftime("%m/%d/%Y"))
```

# Next 90 Days

```python
ninety_d = today + datetime.timedelta(days=90)
```

```python
print("90 days from today is " + ninety_d.strftime("%B %d, %Y"))
```

# Time for Dinner

```python
dinner_time = datetime.time(7, 30)
```

```python
print("Let’s meet for dinner at " + dinner_time.strftime("%I:%M %p"))
```
