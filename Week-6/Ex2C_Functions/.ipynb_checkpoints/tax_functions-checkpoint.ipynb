# Tax Functions Lab

This is Lab 2 of Exercise 2.C. In this notebook, I am creating custom Python functions to calculate payroll taxes, including Social Security tax, Medicare tax, and federal tax withholding based on a withholding code.

```python
def get_soc_sec_tax(gross_pay):
    tax = gross_pay * 0.062
    return tax
```

```python
def get_medicare_tax(gross_pay):
    tax = gross_pay * 0.0145
    return tax
```

```python
def get_federal_tax(gross_pay, withholding_code):
    if withholding_code == 0:
        tax_rate = 0.23
    elif withholding_code == 1:
        tax_rate = 0.21
    elif withholding_code == 2:
        tax_rate = 0.195
    elif withholding_code == 3:
        tax_rate = 0.185
    else:
        tax_rate = 0.18 - ((withholding_code - 4) * 0.005)

    federal_tax = gross_pay * tax_rate
    return federal_tax
```

```python
gross_pay = 750
withholding_code = 0

print("Person 1")
print(f"Gross Pay: ${gross_pay:.2f}")
print(f"Social Security Tax: ${get_soc_sec_tax(gross_pay):.2f}")
print(f"Medicare Tax: ${get_medicare_tax(gross_pay):.2f}")
print(f"Federal Tax: ${get_federal_tax(gross_pay, withholding_code):.2f}")
```
