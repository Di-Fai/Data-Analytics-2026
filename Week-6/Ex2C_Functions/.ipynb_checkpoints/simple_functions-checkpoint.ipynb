This is Lab 1 of Exercise 2.C. In this notebook, I am working with defining custom functions in Python.

```python
def favorite_things(name, movie, music):
    return f"My name is {name}.\nMy favorite movie is {movie}.\nMy favorite musician/band is {music}."

print(favorite_things("Dimitri Nji", "Inception", "Eric Bellinger"))
```

```python
def why_im_here(purpose):
    return purpose

print(why_im_here("I joined the Year Up United Data Analyst program to build stronger technical skills, grow professionally, and create better career opportunities in technology."
))
```

```python
def favorite_place(place, why):
    return f"One of my favorite places is {place} because {why}."
```

```python
print(favorite_place("the beach", "it helps me relax and clear my mind"))
```

```python
print(favorite_place("the gym", "it helps me stay focused, disciplined, and healthy"))
```

For this lab, I created each function with clear parameters so the function can be reused with different information. The most efficient function was why_im_here() because it only returns one sentence. The most readable and reusable function was favorite_place() because another coder can easily change the place and reason without rewriting the full sentence.
