## Lab 3. Employee data
This lab exercise makes updates to information extracted from
an employee database

```python
hr_list = [
    ('0123', 'HR', 'Rebecca Yang', '69000'),
    ('0121', 'IT', 'Mark Blick', '67000'),
    ('0068', 'IT', 'Kahna Larsen', '112000'),
    ('0234', 'OPS', 'Jim Smith', '54000')
]
```

```python
# Mark is the second employee in the list, so his index is 1.
# Since tuples cannot be changed directly, we replace the whole tuple.

hr_list[1] = ('0121', 'IT', 'Mark Blick-Hawley', '67000')
```

```python
print(hr_list)
```

```python
for employee in hr_list:
    employee_number = employee[0]
    dept_code = employee[1]
    name = employee[2]
    salary = int(employee[3])

    print(f"{employee_number} | {dept_code} | {name} | {salary:,}")
```
