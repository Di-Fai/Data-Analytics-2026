```python
import random
```

```python
# 1. random.randint(a, b) to Random interger between a b(inclusive)
print("randint(1, 10):", random.randint(1, 10))
```

```python
# 2. random.random(0 to Random float between 0.0 and 1.0
print("random():", random.random())
```

```python
# 3. random.choice(seq) -> Random single element from a sequence
fruits = ["apple", "banana", "cherry"]
print("choice(fruits):", random.choice(fruits))
```

```python
# 4. random.choices(seq, k=n) -> List of k random elements with replacement
print("choices(fruits, k=3):", random.choices(fruits, k=3))
```

```python
# 5. random.sample(seq, k=n) -> List of k unique random elements without replacement
print("sample(fruits, k=2):", random.sample(fruits, k=2))
```

```python
# 6. random.shuffle(seq) -> Shuffles a list in place
numbers = [1, 2, 3, 4, 5]
random.shuffle(numbers)
print("shuffled numbers:", numbers)
```

```python
from datetime import date, time, datetime, timedelta, timezone
```

```python
# date(year, month, day)
today = date(2026, 5, 13)
print("Date:", today)
```

```python
# time(hour, minute, second, microsecond)
current_time = time(14, 30, 45)
print("Time:", current_time)
```

```python
# datetime(year, month, day, hour, minute, second)
current_datetime = datetime(2026, 5, 9, 14, 30)
print("Datetime:", current_datetime)
```

```python
# timedelta(days, hours, minutes, weeks, etc.)
future_date = today + timedelta(days=7)
print("Future Date:", future_date)
```

```python
# timezone(timedelta(hours=offset))
est = timezone(timedelta(hours=-5))
```

```python
# datetime.now(timezone)
utc_time = datetime.now(timezone.utc)
print("UTC Time:", utc_time)
```

```python
import statistics
```

```python
# Sample dataset
data = [10, 20, 20, 40, 50]
```

```python
print("=== Python Statistics Module Demo ===\n")
```

```python
# mean() -> average value
print("mean:", statistics.mean(data))
```

```python
# median() -> middle value
print("median:", statistics.median(data))
```

```python
# mode() -> most common value
print("mode:", statistics.mode(data))
```

```python
# stdev() -> standard deviation (sample)
print("stdev:", statistics.stdev(data))
```

```python
# variance() -> variance (sample)
print("variance:", statistics.variance(data))
```
