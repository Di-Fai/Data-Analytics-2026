### multiple parameters
### Python automatically packages multiple return values into a tuple

```python
def greeting(name, city, hobby):
    return name, city, hobby

name = input("Please enter your name: ")
city = input("Please enter your city: ")
hobby = input("Please enter your hobby: ")
result = greeting(name, city, hobby)

print(f"Hello, {result[0]}! I see you are from {result[1]} and enjoy {result[2]}.")
```

## use the default parameter city = New York

```python
def greeting(name, hobby, city="New York"):
    return f"Hello, {name}! you are from {city} and you enjoy {hobby}."
```

```python
def add_numbers(num1, num2):
    total = num1 + num2
    return total

num1 = int(input("enter first number"))
num2 = int(input("enter second number"))
    
result = add_numbers(num1, num2)

print(f"The total is {result}")
```

```python
# 1. Creating a string
text = "Python"
```

```python
# 2. Indexing: getting one character from the string
print("First character:", text[0])     
print("Last character:", text[-1])     
```

```python
# 3. Slicing: getting part of the string
print("First three characters:", text[0:3])      
print("From index 2 to end:", text[2:])          
print("Every second character:", text[::2])      
```

```python
# 4. Iteration: going through each character in the string
print("Characters in string:")
for char in text:
    print(char, end=" ")
print()
```

```python
# 5. Membership testing: checking if something exists inside the string
print("'Py' in text?", "Py" in text)          
print("'Java' in text?", "Java" in text)      
```

```python
# 6. Length of string
print("Length of string:", len(text))         
```

```python
# 7. Concatenation and repetition
print("Concatenation:", text + "3.11")        
print("Repetition:", text * 2) 
```

```python
# 8. Immutability demonstration
try:
    text[0] = "J"    
except TypeError as e:
    print("Strings are immutable:", e)
```

```python
def split_sentence(sentence):
    words = sentence.split()
    return words


user_sentence = input("Enter a sentence: ")

result = split_sentence(user_sentence)

print(f"The split sentence is: {result}")
```
